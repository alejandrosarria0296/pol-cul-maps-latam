```python
import os
import pandas as pd
```

```python
file_path_span = r"../president_csvs/5k"
file_path_coca = r"../president_csvs/coca"

for df in os.listdir(file_path_span):
    name = df.split(".")[0]
    pres_df = pd.read_csv(file_path_span + "/" + df)
    sample = pres_df.sample(50)
    sample.to_csv(r"../president_csvs/samples/" + name + "_span_sample.csv", index=False)

for df in os.listdir(file_path_coca):
    name = df.split(".")[0]
    pres_df = pd.read_csv(file_path_coca + "/" + df)
    sample = pres_df.sample(15)
    sample.to_csv(r"../president_csvs/samples/" + name + "_coca_sample.csv", index=False)
```
